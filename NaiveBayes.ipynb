```python
# Importing Required Libraries
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score
```

```python
# Step 1: Load the Dataset
msg = pd.read_csv('document.csv', names=['message', 'label'])

msg.shape
# msg.shape[1]
```

```python
# Step 2: Convert Labels into Numerical Values
msg['labelnum'] = msg.label.map({'pos': 1, 'neg': 0})

# Step 3: Separate Input and Output
X = msg.message
y = msg.labelnum
```

```python
# Step 4: Split Dataset into Training and Testing Data
Xtrain, Xtest, ytrain, ytest = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Step 5: Convert Text into Numerical Features
count_v = CountVectorizer()

Xtrain_dm = count_v.fit_transform(Xtrain)
Xtest_dm = count_v.transform(Xtest)
```

```python
# # Step 6: Display the Numerical Representation of Text
# df = pd.DataFrame(
#     Xtrain_dm.toarray(),
#     columns=count_v.get_feature_names_out()
# )

# print("\nNumerical Representation of Training Data:")
# print(df)
```

```python
# Step 7: Create and Train Naive Bayes Classifier
clf = MultinomialNB()    #  Creates a Multinomial Naïve Bayes classifier.

clf.fit(Xtrain_dm, ytrain)


# Step 8: Predict the Results
pred = clf.predict(Xtest_dm)
```

```python
# Step 9: Evaluate Model Performance

print("Accuracy: ", accuracy_score(ytest, pred))
print("Recall: ", recall_score(ytest, pred))
print("Precision: ", precision_score(ytest, pred))
print("Confusion Matrix:\n", confusion_matrix(ytest, pred))
```
